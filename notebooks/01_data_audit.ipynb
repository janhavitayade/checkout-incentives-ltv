```python
# ============================================================
# ORDER-LEVEL FINANCIAL CONSISTENCY
# ============================================================

# Calculate total order value from order items
item_totals = (
    order_items
    .groupby("order_id")
    .agg(
        item_value=("price", "sum"),
        freight_value=("freight_value", "sum")
    )
)

item_totals["order_item_total"] = (
    item_totals["item_value"] +
    item_totals["freight_value"]
)

# Calculate total payment value per order
payment_totals = (
    payments
    .groupby("order_id")["payment_value"]
    .sum()
    .rename("payment_total")
)

# Combine both
financial_check = item_totals.join(
    payment_totals,
    how="outer"
)

# Difference
financial_check["difference"] = (
    financial_check["payment_total"] -
    financial_check["order_item_total"]
)

financial_check["absolute_difference"] = (
    financial_check["difference"].abs()
)

print("============================================================")
print("FINANCIAL CONSISTENCY")
print("============================================================")

print("Orders with item records:",
      item_totals.shape[0])

print("Orders with payment records:",
      payment_totals.shape[0])

print("Orders with both:",
      financial_check.dropna(
          subset=["order_item_total", "payment_total"]
      ).shape[0])

print()
print("Exact matches:",
      (financial_check["absolute_difference"] < 0.01).sum())

print("Differences > 0.01:",
      (financial_check["absolute_difference"] >= 0.01).sum())

print()
print("Maximum absolute difference:",
      financial_check["absolute_difference"].max())

print("Mean absolute difference:",
      financial_check["absolute_difference"].mean())

print("Median absolute difference:",
      financial_check["absolute_difference"].median())
```

```python
# ============================================================
# LARGEST FINANCIAL DISCREPANCIES
# ============================================================

largest_differences = (
    financial_check
    .dropna(subset=["order_item_total", "payment_total"])
    .sort_values("absolute_difference", ascending=False)
    .head(20)
)

largest_differences[
    [
        "item_value",
        "freight_value",
        "order_item_total",
        "payment_total",
        "difference",
        "absolute_difference"
    ]
]
```

```python
# ------------------------------------------------------------
# CHECK MISSING TIMESTAMPS BY ORDER STATUS
# ------------------------------------------------------------
# We want to understand whether missing timestamps are expected
# for certain order statuses.
#
# For each status, we calculate:
#   missing values / total orders with that status
# ------------------------------------------------------------

timestamp_columns = [
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
]

# Group orders by their status
status_missing = orders.groupby("order_status").agg(
    total_orders=("order_id", "count"),
    approved_missing=("order_approved_at", lambda x: x.isna().sum()),
    carrier_missing=("order_delivered_carrier_date", lambda x: x.isna().sum()),
    customer_delivery_missing=("order_delivered_customer_date", lambda x: x.isna().sum())
)

# Convert the missing counts into "missing / total" strings
status_missing["approved_missing_ratio"] = (
    status_missing["approved_missing"].astype(str)
    + "/"
    + status_missing["total_orders"].astype(str)
)

status_missing["carrier_missing_ratio"] = (
    status_missing["carrier_missing"].astype(str)
    + "/"
    + status_missing["total_orders"].astype(str)
)

status_missing["customer_delivery_missing_ratio"] = (
    status_missing["customer_delivery_missing"].astype(str)
    + "/"
    + status_missing["total_orders"].astype(str)
)

print(status_missing)
```

```python
# ------------------------------------------------------------
# CUSTOMER STATE → ORDER → SELLER ANALYSIS
# ------------------------------------------------------------
# Goal:
# Determine whether states with many customers also generate
# many orders and sales.
#
# We connect:
# customers → orders → order_items → sellers
# ------------------------------------------------------------

# Keep the customer information needed for the analysis
customer_order = orders.merge(
    customers[
        ["customer_id", "customer_unique_id", "customer_state"]
    ],
    on="customer_id",
    how="left"
)

# Add item and seller information
customer_sales = customer_order.merge(
    order_items[
        ["order_id", "product_id", "seller_id", "price", "freight_value"]
    ],
    on="order_id",
    how="left"
)

# Add seller state
customer_sales = customer_sales.merge(
    sellers[
        ["seller_id", "seller_state"]
    ],
    on="seller_id",
    how="left"
)

# Display the requested columns
customer_sales[
    [
        "customer_unique_id",
        "customer_state",
        "order_id",
        "price",
        "seller_id",
        "seller_state"
    ]
].head(20)
```

```python
# ------------------------------------------------------------
# SUMMARIZE CUSTOMER ACTIVITY BY CUSTOMER STATE
# ------------------------------------------------------------
# We compare:
#   - registered customers
#   - customers who actually placed orders
#   - number of orders
#   - sales value
#   - number of sellers involved
# ------------------------------------------------------------

state_summary = customer_sales.groupby("customer_state").agg(
    unique_customers=("customer_unique_id", "nunique"),
    orders=("order_id", "nunique"),
    sales_value=("price", "sum"),
    unique_sellers=("seller_id", "nunique")
).sort_values("orders", ascending=False)

print(state_summary)
```

```python
# ------------------------------------------------------------
# CHECK REVIEW TIMELINE
# ------------------------------------------------------------
# Goal:
# Determine whether reviews were created after the relevant
# order events occurred.
#
# We join reviews with orders using order_id.
# ------------------------------------------------------------

review_timeline = reviews.merge(
    orders[
        [
            "order_id",
            "order_status",
            "order_purchase_timestamp",
            "order_approved_at",
            "order_delivered_carrier_date",
            "order_delivered_customer_date",
            "order_estimated_delivery_date"
        ]
    ],
    on="order_id",
    how="left"
)

# Select the columns needed to inspect the chronology
review_timeline = review_timeline[
    [
        "order_id",
        "order_status",
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "review_creation_date",
        "review_answer_timestamp",
        "review_score"
    ]
]

# Display some records
print(review_timeline.head(20))
```

```python
# ------------------------------------------------------------
# CHECK FOR CHRONOLOGICAL INCONSISTENCIES IN REVIEWS
# ------------------------------------------------------------

# Review created before purchase
review_before_purchase = (
    review_timeline["review_creation_date"]
    < review_timeline["order_purchase_timestamp"]
).sum()

# Review created before approval
review_before_approval = (
    review_timeline["review_creation_date"]
    < review_timeline["order_approved_at"]
).sum()

# Review created before carrier/shipment date
review_before_shipment = (
    review_timeline["review_creation_date"]
    < review_timeline["order_delivered_carrier_date"]
).sum()

# Review created before customer delivery
review_before_delivery = (
    review_timeline["review_creation_date"]
    < review_timeline["order_delivered_customer_date"]
).sum()

print("Review before purchase:", review_before_purchase)
print("Review before approval:", review_before_approval)
print("Review before shipment:", review_before_shipment)
print("Review before delivery:", review_before_delivery)
```

```python
# ------------------------------------------------------------
# CHECK WHETHER REVIEWS ARE ASSOCIATED WITH DELIVERED ORDERS
# ------------------------------------------------------------

review_status = (
    review_timeline
    .groupby("order_status")
    .size()
    .sort_values(ascending=False)
)

print(review_status)
```
